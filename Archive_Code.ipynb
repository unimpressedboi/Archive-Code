```python
def pdf_to_image(pdf_path):
  doc = fitz.open(pdf_path)
  page_1 = doc.load_page(0)
  zoom = 4.0
  zoom_mat = fitz.Matrix(zoom, zoom)
  pix = page_1.get_pixmap(matrix = zoom_mat)
  pix.save("output.jpg")

```
